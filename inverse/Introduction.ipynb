```python
# Import earthkit-data for accessing sample datasets
import earthkit.data as ekd
# Import earthkit-plots to visualise them
import earthkit.plots as ekp

# Where are the meteorological data files located for the FLEXPART inversion example?
inputdir = "???"  # Replace with the actual path to the meteorological data files
```

```python
# What files or dates are available in the input directory?

# format: YYYYMMDDHH
date = "???" # Replace with the actual date of the meteorological data file you want to read

# This opens the meteorological data file.
ds_in = ekd.from_source("file", f"{inputdir}/EA{date}")

# What is inside these meteorological data files, and how are they formatted?
ds = ds_in.to_fieldlist()
```

# Plotting the input data


## Surface fields

Here we look at surface fields, e.g. 2m temperature and mean sea level pressure. We can just select these variables and plot the data right away.

```python
# Select the temperature and pressure fields from the dataset. What are the units of these fields?
temperature = ds.sel({"parameter.variable": "2t"})
pressure = ds.sel({"parameter.variable": "msl"})

# Create a map
chart = ekp.Map()

# ECMWF has defined presets for plotting. Nice!
chart.plot(temperature, units="celsius")
chart.plot(pressure, units="hPa")

# Add a legend to the map. What does the legend show?
chart.legend(location="right")
# Plot the location of the FLEXPART receptor.
chart.annotate("+", (25.6,35.34),color="r")
chart.coastlines() # add coastlines 
chart.title() # add a title
chart.gridlines() # add gridlines
chart.show() # show
```

or using the wind fields at 10m height. The wind fields are given as u and v components, which can be combined to get the wind speed and direction.

```python
# winds
chart = ekp.Map()
chart.quiver(ds.sel({"parameter.variable": ["10u","10v"]}), cmap="turbo")
chart.plot(ds.sel({"parameter.variable": ["msl"]}), units="hPa")
chart.legend(label="wind speed ({units})", location="right")
# Plot the location of the FLEXPART receptor.
chart.annotate("+", (25.6,35.34),color="r")
chart.coastlines()
chart.land()
chart.gridlines()
chart.title()
chart.show()
```

## level fields

We can also use the 3d fields, e.g. winds on 850 hPa level. The wind fields are given as u and v components, which can be combined to get the wind speed and direction.

Because the vertical is using hybrid coordinates, which is a combination of pressure and sigma coordinates, we can also select the vertical level by its index. For example, the 850 hPa level is the 114 level in the vertical or 500 hPa is the 96 level.



```python
# winds at 850 hPa level
chart = ekp.Map()
winds = ds.sel({"parameter.variable": ["u","v"], "vertical.level": 114})
chart.quiver(winds, cmap="turbo")
chart.streamplot(winds)
chart.legend(label="wind speed ({units})", location="right")
chart.annotate("+", (25.6, 35.34),color="r")
chart.coastlines()
chart.land()
chart.gridlines()
chart.title()
chart.show()
```
